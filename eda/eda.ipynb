## Notes on data quality

- `Latitude`/`Longitude` are stored as text (e.g. "6.0N", "33.3W").
- `Calendar Date` has negative years for BCE (e.g. "-1999 June 12") so --> extract the year separately.
- `Path Width (km)` and `Central Duration` are missing 4200 rows.

```python
# Check for missing values per column
df.isna().sum()
```

`Path Width` and `Central Duration` missing is to be expected, since only central eclipses have a path on Earth.

## Distribution of the many eclipse types

```python
# First letter of the eclipse type indicates the type of eclipse. Map that to a more descriptive name.
type_names = {"P": "Partial", "A": "Annular", "T": "Total", "H": "Hybrid"}
df["Type Name"] = df["Eclipse Type"].str[0].map(type_names).fillna("Other")
df["Type Name"].value_counts()
```

```python
df["Type Name"].value_counts().plot(
    kind="bar", title="Number of solar eclipses per type", ylabel="Count"
)
```

## Time/Years

```python
df["Year"] = df["Calendar Date"].str.strip().str.split().str[0].astype(int)
print(f"From year {df['Year'].min()} to year {df['Year'].max()}")
df["Year"].describe()
```

```python
# Eclipses per century
((df["Year"] // 100) * 100).value_counts().sort_index().plot(
    kind="line", title="Solar eclipses per century", xlabel="Century", ylabel="Count"
)
```

## Magnitude

```python
# Magnitude > 1 = total, ~1 = annular/hybrid, < 1 = partial
df["Eclipse Magnitude"].plot(
    kind="hist", bins=40, title="Distribution of eclipse magnitude"
)
```
